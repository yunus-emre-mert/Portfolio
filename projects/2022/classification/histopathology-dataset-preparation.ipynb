```python
import numpy as np
import pandas as pd
import os
import tensorflow as tf
import matplotlib.pyplot as plt
import matplotlib.image as mpimg
import warnings
import matplotlib.pyplot as plt
from sklearn import model_selection
from sklearn.datasets import load_iris
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score
from sklearn.metrics import confusion_matrix

import cv2
import random
```

```python
from keras.preprocessing.image import ImageDataGenerator
from keras.callbacks import ReduceLROnPlateau
from keras.models import Sequential
from keras.layers import Dense, Dropout, Flatten, Conv2D, MaxPool2D
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.preprocessing import image
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.optimizers import RMSprop
from keras import regularizers
from keras.layers.core import Dropout
from keras.applications.vgg16 import VGG16
```

```python
train = '/content/CNN DATA/train'
test =  '/content/CNN DATA/test'

classes = ["adenosis","fibroadenoma","papillary_carcinoma","tubular_adenoma","ductal_carcinoma","lobular_carcinoma","mucinous_carcinoma","papillary_carcinoma"]
```

```python
train_data = ImageDataGenerator(rescale=1./255,
                                   zoom_range=0.3,validation_split=0.15,
                                   horizontal_flip=True)
test_data = ImageDataGenerator(rescale=1./255)



traindata = train_data.flow_from_directory(train,
                                                batch_size=8,subset = "training",
                                                target_size=(224,224),
                                                shuffle=True,
                                                color_mode='rgb',
                                                class_mode='categorical')

validationdata = train_data.flow_from_directory(train,
                                                batch_size=8,
                                                target_size=(224,224),subset = "validation",
                                                shuffle=True,
                                                color_mode='rgb',
                                                class_mode='categorical')

testdata = test_data.flow_from_directory(test,
                                                batch_size=8,
                                                target_size=(224,224),
                                                shuffle=False,
                                                color_mode='rgb',
                                                class_mode='categorical')
```

```python
vgg = VGG16(weights='imagenet',
              include_top=False,
              input_shape = (224,224,3))

for layer in vgg.layers:
    layer.trainable = False

vgg16Model = Sequential()
vgg16Model.add(vgg)
vgg16Model.add(Flatten())
vgg16Model.add(Dense( 8, activation = "softmax"))

vgg16Model.summary()
```

```python
steps_per_epoch = traindata.n // traindata.batch_size
validation_steps = validationdata.n // validationdata.batch_size
```

```python
vgg16Model.compile(optimizer="adam",loss = "categorical_crossentropy",metrics = ["accuracy"])

hist = vgg16Model.fit(traindata,steps_per_epoch=validation_steps,
                 validation_data=validationdata,validation_steps=validation_steps,
                 epochs=50)
```

```python
score_test_vgg16Model = vgg16Model.evaluate(testdata, steps = len(testdata), verbose = 0)
print('Test loss:', score_test_vgg16Model[0])
print('Test accuracy:', score_test_vgg16Model[1])
```
