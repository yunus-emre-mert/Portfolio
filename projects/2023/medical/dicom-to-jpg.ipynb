```python
!pip install pydicom
import numpy as np
from PIL import Image
import pydicom
```

```python
import glob
dcm_list = glob.glob("/content/Training_Veri/*/*/*/*.dcm")
```

```python
dcm_list[0]
```

```python
dcm_list[0].split('.dcm')[0].split('/')[6]
```

```python
for ds in dcm_list:
  print(ds.split('.dcm')[0].split('/')[6])
```

```python
for ds in dcm_list:
  filename = ds.split('.dcm')[0].split('/')[6]
  ds = pydicom.dcmread(ds)
  new_image = ds.pixel_array.astype(float)
  scaled_image = (np.maximum(new_image, 0) / new_image.max()) * 255.0
  scaled_image = np.uint8(scaled_image)
  final_image = Image.fromarray(scaled_image)
  final_image.show()
  final_image.save("/content/jpg/"+ filename + '.jpg')
  
```
