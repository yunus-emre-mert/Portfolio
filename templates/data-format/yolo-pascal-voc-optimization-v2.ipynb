```python
import shutil 
from glob import glob
import os 
```

```python
os.mkdir("/content/alldata")
os.mkdir("/content/Faster rcnn data dermo")
os.mkdir("/content/Faster rcnn data dermo/train")
os.mkdir("/content/Faster rcnn data dermo/test")
```

```python
images = glob("/content/content/datasets/images/*/*.jpg")
pascal_labels = glob("/content/content/jsonlabels/*.json")
print("labels",len(pascal_labels))
print("images",len(images))
```

```python
for i in images:
  shutil.copy(i,"/content/alldata")
for a in pascal_labels:
  shutil.copy(a,"/content/alldata")
```

```python
from sklearn.model_selection import train_test_split

images = glob("/content/alldata/*.jpg")
pascal_labels = glob("/content/alldata/*.json")
print(len(pascal_labels))
print(len(images))


train_labels, test_labels= train_test_split(pascal_labels,shuffle=True, test_size=0.25, random_state=42)
print("train labels count",len(train_labels))
print("test labels count",len(test_labels))


```

```python
for i in train_labels:
  shutil.copy(i,"/content/Faster rcnn data dermo/train")
for a in test_labels:
  shutil.copy(a,"/content/Faster rcnn data dermo/test")
```
